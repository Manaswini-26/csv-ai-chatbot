## Turn Any CSV into an AI Chatbot with Python

### Step 1: Loading and Serializing the CSV

```python
import pandas as pd
df=pd.read_csv("C:/Users/kalya/OneDrive/ドキュメント/sensex.csv")
df
```

```python
documents=df.astype(str).apply( lambda row:" | ".join(row),axis=1).tolist()
print(documents[:3])
```

### Step 2: Generating Local Embeddings

```python
from sentence_transformers import SentenceTransformer
model=SentenceTransformer("all-MiniLM-L6-v2")
embeddings=model.encode(documents)
```

### Step 3: Indexing with FAISS

```python
import faiss
import numpy as np

dimension=embeddings.shape[1]

index=faiss.IndexFlatL2(dimension)
index.add(np.array(embeddings))
```

### Step 4: Querying the Data

```python
query='What was the closing stock on 2025-03-27?'
query_vector=model.encode([query])
D,I=index.search(query_vector,k=3)
results=[documents[i] for i in I[0]]
print(results)
```
